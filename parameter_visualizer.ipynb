## Minimization analysis

```python
# Parallel case (statistically speaking, close to SM-like)
parall_case = len(Z2_violating_bests_parall) / (len(Z2_violating_bests_parall) + len(Z2_violating_bests_perp)) * 100

# Perpendicular case (statistically speaking, the purest 2HDM case)
perp_case = len(Z2_violating_bests_perp) / (len(Z2_violating_bests_parall) + len(Z2_violating_bests_perp)) * 100
```

# Symmetric and antisymmetric 2HDM potentials

```python
# Select type (symmetric or antisymmetric)
type = "antisymmetric"

# Advanced settings
font_size = 50
ax_lim = 10
cmap = 'spring'
parameter_file_name = "Z2_violating_antisymmetric_parameters.txt"
best_file_name = "Z2_violating_antisymmetric_bests.txt"

match type:
    case "antisymmetric":
        antisymmetric_valid_parameters = np.loadtxt(parameter_file_name, delimiter=',')
        antisymmetric_valid_bests = np.loadtxt(best_file_name, delimiter=',')

        plt.rcParams['xtick.labelsize'] = 20
        plt.rcParams['ytick.labelsize'] = 20

        # Define effective parameters
        rho_3_Z_2_violating_antisym = antisymmetric_valid_parameters[:,0]
        rho_4_Z_2_violating_antisym = antisymmetric_valid_parameters[:,1]
        rho_5_Z_2_violating_antisym = antisymmetric_valid_parameters[:,2]
        alpha_5_Z_2_violating_antisym = antisymmetric_valid_parameters[:,3]
        rho_6_Z_2_violating_antisym = antisymmetric_valid_parameters[:,4]
        alpha_6_Z_2_violating_antisym = antisymmetric_valid_parameters[:,5]
        rho_7_Z_2_violating_antisym = antisymmetric_valid_parameters[:,6]
        alpha_7_Z_2_violating_antisym = antisymmetric_valid_parameters[:,7]
        r_antisym = antisymmetric_valid_bests[:,0]
        phi_antisym = antisymmetric_valid_bests[:,1]

        rho_3_eff_Z_2_violating_antisym = rho_3_Z_2_violating_antisym + r_antisym**2*(rho_4_Z_2_violating_antisym + rho_5_Z_2_violating_antisym*np.cos(alpha_5_Z_2_violating_antisym + phi_antisym))

        rho_6_eff_Z_2_violating_antisym = rho_6_Z_2_violating_antisym*r_antisym*np.cos(alpha_6_Z_2_violating_antisym + (1/2)*phi_antisym)

        rho_7_eff_Z_2_violating_antisym = rho_7_Z_2_violating_antisym*r_antisym*np.cos(alpha_7_Z_2_violating_antisym + (1/2)*phi_antisym)

        u_Z_2_violating_antisym = rho_6_eff_Z_2_violating_antisym + rho_7_eff_Z_2_violating_antisym

        v_Z_2_violating_antisym = rho_6_eff_Z_2_violating_antisym - rho_7_eff_Z_2_violating_antisym

        plt.figure(figsize=(10,10))

        var_x = np.linspace(-1,25,1000)
        var_x2 = np.linspace(-1,25,1000)

        plt.plot(var_x,var_x, color='black',linestyle=':', linewidth=3, label=r'$|v_{\text{Vmin}}|=\rho^{eff}_{3,\text{Vmin}}+3$',zorder=2)
        plt.plot(var_x,(var_x**2-8)/8+3, color='mediumblue', linewidth=3, label=r'$v^2_{\text{Vmin}} = 8(\rho^{eff}_{3,\text{Vmin}}+1)$',zorder=2)
        plt.plot(var_x2,np.linspace(2,2,len(var_x2)), linestyle='-.', linewidth=3 ,color='red',label=r'$\rho^{eff}_{3,\text{Vmin}}=-1$',zorder=2)
        plt.xlim([0,12])
        plt.ylim([-1, 20])

        plt.scatter(np.abs(v_Z_2_violating_antisym),rho_3_eff_Z_2_violating_antisym+3, s=2, color='goldenrod',zorder=1)
        plt.ylabel(r'$\rho^{eff}_{3,\text{Vmin}}+3$',fontsize=30)
        plt.xlabel(r'$|v_{\text{Vmin}}|$',fontsize=30)
        plt.legend(loc='upper center',fontsize=20)
        plt.savefig('antisym_restriction.png', dpi=300, bbox_inches='tight')
        plt.show()

    case "symmetric":
        symmetric_valid_values = np.loadtxt(parameter_file_name, delimiter=',')
        symmetric_valid_bests = np.loadtxt(best_file_name, delimiter=',')

        plt.rcParams['xtick.labelsize'] = 20
        plt.rcParams['ytick.labelsize'] = 20

        # Define effective parameters
        rho_3_Z_2_violating_sym = symmetric_valid_values[:,0]
        rho_4_Z_2_violating_sym = symmetric_valid_values[:,1]
        rho_5_Z_2_violating_sym = symmetric_valid_values[:,2]
        alpha_5_Z_2_violating_sym = symmetric_valid_values[:,3]
        rho_6_Z_2_violating_sym = symmetric_valid_values[:,4]
        alpha_6_Z_2_violating_sym = symmetric_valid_values[:,5]
        rho_7_Z_2_violating_sym = symmetric_valid_values[:,6]
        alpha_7_Z_2_violating_sym = symmetric_valid_values[:,7]
        r_sym = symmetric_valid_bests[:,0]
        phi_sym = symmetric_valid_bests[:,1]
        
        rho_3_eff_Z_2_violating_sym = rho_3_Z_2_violating_sym + r_sym**2*(rho_4_Z_2_violating_sym + rho_5_Z_2_violating_sym*np.cos(alpha_5_Z_2_violating_sym + phi_sym))

        rho_6_eff_Z_2_violating_sym = rho_6_Z_2_violating_sym*r_sym*np.cos(alpha_6_Z_2_violating_sym + (1/2)*phi_sym)

        rho_7_eff_Z_2_violating_sym = rho_7_Z_2_violating_sym*r_sym*np.cos(alpha_7_Z_2_violating_sym + (1/2)*phi_sym)

        u_Z_2_violating_sym = rho_6_eff_Z_2_violating_sym + rho_7_eff_Z_2_violating_sym

        v_Z_2_violating_sym = rho_6_eff_Z_2_violating_sym - rho_7_eff_Z_2_violating_sym

        limits = [-0.25, 10]
        y_limits = [-2.5, 16]
        x_var = np.linspace(0,4,1000)
        x_var2 = np.linspace(4,limits[1],1000)
        x_var3 = np.linspace(limits[0],limits[1],1000)
        x_var4 = np.linspace(0,limits[1],1000)
        x_var5 = np.linspace(y_limits[0],y_limits[1],1000)
        aux_var = np.linspace(0,0,1000)

        plt.figure(figsize=(10,10))

        plt.scatter(-u_Z_2_violating_sym,rho_3_eff_Z_2_violating_sym+1, s=5, color='goldenrod', zorder=2)

        plt.fill_between(x_var4, x_var4/2+2, y_limits[1], color='gray', alpha=0.5, zorder=1, label=r'$u_{\text{Vmin}}=2(1-\rho^{eff}_{3,\text{Vmin}})$, $\rho^{eff}_{3,\text{Vmin}}>1$')
        plt.plot(x_var3,x_var3,color='blue',linestyle='-', linewidth=3, label=r'$\rho^{eff}_{3,\text{Vmin}}+u_{\text{Vmin}}=-1$', zorder=3)
        plt.plot(x_var2,x_var2**2/8+2, color='red',linestyle='-.', linewidth=3, label=r'$8(\rho^{eff}_{3,\text{Vmin}}-1) = u^2_{\text{Vmin}}$', zorder=3)
        plt.plot(x_var4,x_var4/2+2, color='black',linestyle=':', linewidth=3, zorder=3)
        plt.plot(aux_var,x_var5, color='black',linestyle='-', linewidth=3, zorder=3, label=r"$u_{\text{Vmin}}\leq 0$")

        plt.legend(loc=4,fontsize=19)
        plt.ylabel(r'$\rho^{eff}_{3,\text{Vmin}}+1$',fontsize=30)
        plt.xlabel(r'$-u_{\text{Vmin}}$',fontsize=30)
        plt.xlim(limits)
        plt.ylim(y_limits)
        plt.savefig('sym_restriction.png', dpi=300, bbox_inches='tight')
        plt.show()
```

# Relation between the $\rho$'s and physical parameters in the 2HDM $Z_{2}$-conserving case

```python
cba, MH, m, MHp, Y2, MA, beta = sp.symbols('cba, MH, m, MHp, Y2, MA, beta', real=True) # cba is cos(beta-alpha), tan(beta) = v2/v1, Y2 is a cuadratic term and the rest are masses.
```

```python
Z1 = (1-cba**2)*(MH**2*cba**2/(1-cba**2)+m**2)
Z3 = 2*MHp**2-2*Y2
Z4 = 1/2*(2*cba**2-1)*(m**2-MH**2)+m**2/2+MA**2+MH**2/2-2*MHp**2
Z5 = 1/2*(2*cba**2-1)*(m**2-MH**2)+m**2/2-MA**2+MH**2/2
Z6 = sp.sqrt(1-cba**2)*cba*(m**2-MH**2)

Z345 = Z3 + Z4 + sp.re(Z5)

Z2 = 1/(32*sp.cos(beta)**3*sp.sin(beta)**3)*(-8*Z6*(sp.cos(2*beta)+sp.cos(6*beta))+sp.sin(2*beta)*(Z1+2*Z345)+sp.sin(6*beta)*(2*Z345-3*Z1))
Z7 = 1/(8*sp.cos(beta)**2*sp.sin(beta)**2)*(-3*sp.cos(4*beta)*Z6+sp.sin(4*beta)*(-Z1+Z3+Z4+Z5)-Z6)
```

```python
lambda_1 = Z1*sp.cos(beta)**4+Z2*sp.sin(beta)**4+1/2*Z345*sp.sin(2*beta)**2+2*sp.sin(2*beta)*(sp.cos(beta)**2*sp.re(Z6)+sp.sin(beta)**2*sp.re(Z7))
lambda_2 = Z1*sp.sin(beta)**4+Z2*sp.cos(beta)**4+1/2*Z345*sp.sin(2*beta)**2-2*sp.sin(2*beta)*(sp.sin(beta)**2*sp.re(Z6)+sp.cos(beta)**2*sp.re(Z7))
lambda_3 = Z3+1/4*sp.sin(2*beta)**2*(Z1+Z2-2*Z345)-sp.sin(2*beta)*sp.cos(2*beta)*sp.re(Z6-Z7)
lambda_4 = Z4+1/4*sp.sin(2*beta)**2*(Z1+Z2-2*Z345)-sp.sin(2*beta)*sp.cos(2*beta)*sp.re(Z6-Z7)
lambda_5 = 1/4*sp.sin(2*beta)**2*(Z1+Z2-2*Z345)+sp.re(Z5)-sp.sin(2*beta)*sp.cos(2*beta)*sp.re(Z6-Z7)
```

```python
Y2_value = 750 # GeV
vev = 246 # GeV # vev of the SM higgs boson
beta_value = np.pi/4

lambda_1 = lambda_1.subs({m:125/vev, Y2:Y2_value**2/vev**2, MHp:MH, MA:MH})
lambda_2 = lambda_2.subs({m:125/vev, Y2:Y2_value**2/vev**2, MHp:MH, MA:MH})
lambda_3 = lambda_3.subs({m:125/vev, Y2:Y2_value**2/vev**2, MHp:MH, MA:MH})
lambda_4 = lambda_4.subs({m:125/vev, Y2:Y2_value**2/vev**2, MHp:MH, MA:MH})
lambda_5 = lambda_5.subs({m:125/vev, Y2:Y2_value**2/vev**2, MHp:MH, MA:MH})
```

```python
X = np.linspace(-0.9999, 0.9999, 5000)
Y = np.linspace(0, 16, 5000)
x, y = np.meshgrid(X, Y)

lambda_1 = lambda_1.subs({beta:beta_value})
lambda_1_num = sp.lambdify([cba,MH], lambda_1, modules='numpy')
lambda_1 = lambda_1_num(x,y)

lambda_2 = lambda_2.subs({beta:beta_value})
lambda_2_num = sp.lambdify([cba,MH], lambda_2, modules='numpy')
lambda_2 = lambda_2_num(x,y)

lambda_3 = lambda_3.subs({beta:beta_value})
lambda_3_num = sp.lambdify([cba,MH], lambda_3, modules='numpy')
lambda_3 = lambda_3_num(x,y)

lambda_4 = lambda_4.subs({beta:beta_value})
lambda_4_num = sp.lambdify([cba,MH], lambda_4, modules='numpy')
lambda_4 = lambda_4_num(x,y)

lambda_5 = lambda_5.subs({beta:beta_value})
lambda_5_num = sp.lambdify([cba,MH], lambda_5, modules='numpy')
lambda_5 = lambda_5_num(x,y)

rho_3 = lambda_3 / np.sqrt(np.abs(lambda_1*lambda_2))
rho_4 = lambda_4 / np.sqrt(np.abs(lambda_1*lambda_2))
rho_5 = np.abs(lambda_5 / np.sqrt(np.abs(lambda_1*lambda_2)))

rho_3_eff = rho_3 + np.minimum(rho_4-abs(rho_5),0)



# Restrictions

restriction_1 = lambda_1 > 0
restriction_2 = lambda_2 > 0

restriction_3 = rho_3_eff + 1 > 0  # The condition obtanied in the sientific paper.
restriction_4 = lambda_1*lambda_2 - lambda_3**2 > 0  # Positive definite matrix condition.

# Additional restrictions in order to have a positive definite matrix. Due to the fact that these two restrictions might be too strong, we will not include them in the final plot.

# restriction_5 = lambda_5 > 0   
# restriction_6 = lambda_5**2 - lambda_4**2 > 0   

#########################################################################################################################################################################################

# Unitarity conditions


Ginzburg_Lambda_even_21p = np.abs(1/2 * (lambda_1 + lambda_2 + np.sqrt((lambda_1 - lambda_2)**2 + 4*np.abs(lambda_5)**2)))
Ginzburg_Lambda_even_21m = np.abs(1/2 * (lambda_1 + lambda_2 - np.sqrt((lambda_1 - lambda_2)**2 + 4*np.abs(lambda_5)**2)))
Ginzburg_Lambda_even_01p = np.abs(1/2 * (lambda_1 + lambda_2 + np.sqrt((lambda_1 - lambda_2)**2 + 4*lambda_4**2)))
Ginzburg_Lambda_even_01m = np.abs(1/2 * (lambda_1 + lambda_2 - np.sqrt((lambda_1 - lambda_2)**2 + 4*lambda_4**2)))
Ginzburg_Lambda_even_00p = np.abs(1/2 * (3*(lambda_1+lambda_2) + np.sqrt(9*(lambda_1-lambda_2)**2+4*(2*lambda_3+lambda_4)**2)))
Ginzburg_Lambda_even_00m = np.abs(1/2 * (3*(lambda_1+lambda_2) - np.sqrt(9*(lambda_1-lambda_2)**2+4*(2*lambda_3+lambda_4)**2)))
Ginzburg_Lambda_odd_21 = np.abs(lambda_3 + lambda_4)
Ginzburg_Lambda_odd_20 = np.abs(lambda_3 - lambda_4)
Ginzburg_Lambda_odd_01p = np.abs(lambda_3 + np.abs(lambda_5))
Ginzburg_Lambda_odd_01m = np.abs(lambda_3 - np.abs(lambda_5))
Ginzburg_Lambda_odd_00p = np.abs(lambda_3 + 2*lambda_4 + 3*np.abs(lambda_5))
Ginzburg_Lambda_odd_00m = np.abs(lambda_3 + 2*lambda_4 - 3*np.abs(lambda_5))

Ginzburg_restriction = (Ginzburg_Lambda_even_21p < 8*np.pi) & (Ginzburg_Lambda_even_21m < 8*np.pi) & (Ginzburg_Lambda_even_01p < 8*np.pi) & (Ginzburg_Lambda_even_01m < 8*np.pi) & (Ginzburg_Lambda_even_00p < 8*np.pi) & (Ginzburg_Lambda_even_00m < 8*np.pi) & (Ginzburg_Lambda_odd_21 < 8*np.pi) & (Ginzburg_Lambda_odd_20 < 8*np.pi) & (Ginzburg_Lambda_odd_01p < 8*np.pi) & (Ginzburg_Lambda_odd_01m < 8*np.pi) & (Ginzburg_Lambda_odd_00p < 8*np.pi) & (Ginzburg_Lambda_odd_00m < 8*np.pi)


Ginzburg_restrictionII = (Ginzburg_Lambda_even_21p < 16*np.pi) & (Ginzburg_Lambda_even_21m < 16*np.pi) & (Ginzburg_Lambda_even_01p < 16*np.pi) & (Ginzburg_Lambda_even_01m < 16*np.pi) & (Ginzburg_Lambda_even_00p < 16*np.pi) & (Ginzburg_Lambda_even_00m < 16*np.pi) & (Ginzburg_Lambda_odd_21 < 16*np.pi) & (Ginzburg_Lambda_odd_20 < 16*np.pi) & (Ginzburg_Lambda_odd_01p < 16*np.pi) & (Ginzburg_Lambda_odd_01m < 16*np.pi) & (Ginzburg_Lambda_odd_00p < 16*np.pi) & (Ginzburg_Lambda_odd_00m < 16*np.pi)
```
